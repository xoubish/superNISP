# Figure 3 · Conditional diffusion architecture

Implementation-verified vector diagram of the current full diffusion run.
Training, the shared conditional U-Net, and NISP-conditioned generation are
separate panels. This notebook reads source/provenance files only: it never
imports a model, loads checkpoints, launches inference, or uses a GPU.

The colored conditioning paths terminate in the residual-block inset, which
is used twice at every encoder, bottleneck, and decoder stage. No figure title.

```python
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# Coordinates are local to each panel; all labels remain editable in SVG.
INK = '#394553'
BLUE = '#E8EFF7'
TEAL = '#E4F2EC'
PURPLE = '#EEE8F5'
GOLD = '#F7EEDB'
GRAY = '#F1F3F5'
artists_for_review = []

def node(ax, x, y, w, h, text, fill=BLUE, size=7.2, edge='#8B98A5'):
    patch = FancyBboxPatch((x,y),w,h,boxstyle='round,pad=0.025,rounding_size=0.07',
                          facecolor=fill,edgecolor=edge,linewidth=.75,zorder=3)
    ax.add_patch(patch)
    label = ax.text(x+w/2,y+h/2,text,ha='center',va='center',fontsize=size,
                    color=INK,linespacing=1.35,zorder=4)
    artists_for_review.append((patch,label))
    return patch

def link(ax, start, end, color=INK, dashed=False, bend=0, head=True):
    patch = FancyArrowPatch(start,end,arrowstyle='-|>' if head else '-',
        mutation_scale=8,linewidth=.85,color=color,
        linestyle=(0,(3,2)) if dashed else '-',connectionstyle=f'arc3,rad={bend}',
        shrinkA=2,shrinkB=2,zorder=2)
    ax.add_patch(patch)

fig = plt.figure(figsize=FIGSIZE)
gs = fig.add_gridspec(3,1,height_ratios=[1.15,4.15,1.15],hspace=.12,
                      left=.018,right=.985,top=.985,bottom=.015)
train,net,sample = [fig.add_subplot(gs[i]) for i in range(3)]
for ax in [train,net,sample]:
    ax.axis('off'); ax.set_xlim(0,16)
train.set_ylim(0,2.25)
net.set_ylim(0,7.5)
sample.set_ylim(0,2.25)

# (a) Both prediction and reference quantities enter the training objective.
train.text(0,2.13,'(a)  Training',fontsize=9,fontweight='semibold',color=INK)
node(train,.1,.6,2.25,.85,'JWST F115W\ntarget $x_0$',GOLD)
node(train,3.1,.6,3.75,.85,
     r'$x_t=a_t x_0+b_t\epsilon$'+'\nRandom $t$ and noise $\epsilon$',GRAY)
node(train,7.8,.6,3.15,.85,'Conditional U-Net (b)\n'+r'$\hat v_\theta(x_t,t,c)$',BLUE)
node(train,12.1,.6,3.8,.85,
     r'MSE$(\hat v,v)$ + 0.1 L1$(\hat x_0,x_0)$'+'\nValid target pixels only',GOLD,size=6.9)
for start,end in [((2.35,1.025),(3.1,1.025)),((6.85,1.025),(7.8,1.025)),
                  ((10.95,1.025),(12.1,1.025))]:link(train,start,end)
train.text(9.375,1.99,'NISP + validity, $c$',ha='center',fontsize=7.2,color='#007C59')
link(train,(9.375,1.88),(9.375,1.45),color='#007C59')
train.text(8,0.18,r'$a_t=\sqrt{\bar\alpha_t}$,  $b_t=\sqrt{1-\bar\alpha_t}$; '
           r'$v=a_t\epsilon-b_t x_0$,  $\hat x_0=a_t x_t-b_t\hat v$',
           ha='center',fontsize=7,color=INK)

# (b) Feature dimensions are after the two residual blocks at each stage.
net.text(0,7.2,'(b)  Shared conditional U-Net',fontsize=9,fontweight='semibold',color=INK)
net.text(15.9,7.2,'4.91 million parameters',ha='right',fontsize=7.2,color=INK)
node(net,.1,5.9,2.3,.8,'Noisy image $x_t$\n205 × 205 × 1',GRAY)
enc=[(3.2,5.95),(4.15,4.7),(5.1,3.45)]
dec=[(10.6,5.95),(9.65,4.7),(8.7,3.45)]
W,H=2.1,.7
for positions in [enc,dec]:
    for (x,y),dimension in zip(positions,['208² × 32','104² × 64','52² × 128']):
        node(net,x,y,W,H,dimension,BLUE,size=7.4)
node(net,6.65,2.25,3.15,.7,'26² × 128',BLUE,size=7.4)
node(net,13.6,5.9,2.3,.8,'Velocity $\hat v$\n205 × 205 × 1',PURPLE)
link(net,(2.4,6.3),(3.2,6.3))
net.text(2.8,6.83,'pad + conv',ha='center',fontsize=6.2,color=INK)
link(net,(12.7,6.3),(13.6,6.3))
net.text(13.13,6.83,'conv + crop',ha='center',fontsize=6.2,color=INK)
for (x1,y1),(x2,y2) in zip(enc,enc[1:]):
    link(net,(x1+W/2,y1),(x2+W/2,y2+H))
link(net,(enc[-1][0]+W/2,enc[-1][1]),(7.4,2.95))
link(net,(9.05,2.95),(dec[-1][0]+W/2,dec[-1][1]))
for (x1,y1),(x2,y2) in zip(dec[::-1],dec[::-1][1:]):
    link(net,(x1+W/2,y1+H),(x2+W/2,y2))
for (x1,y1),(x2,y2) in zip(enc,dec):
    link(net,(x1+W,y1+H/2),(x2,y2+H/2),color='#8D99A5',dashed=True)
net.text(7.95,6.55,'concatenated skip features',ha='center',fontsize=6.6,color='#657281')
net.text(13.0,3.45,'Down: stride-2 conv\nUp: transposed conv',ha='center',fontsize=6.8,color=INK,linespacing=1.5)
net.text(13.0,2.35,'2 residual blocks per stage\n14 blocks in total',ha='center',fontsize=7,color=INK,linespacing=1.5)

# Condition inputs connect to the modulation inside the shared block inset.
node(net,.1,4.35,2.8,1.0,'NISP Y + validity\n41² → 205²\nCondition pyramid',TEAL,size=7)
node(net,.1,2.9,2.8,.85,'Timestep $t$\n256-feature embedding',PURPLE,size=7)
# A local block inset avoids suggesting that conditioning reaches only one scale.
net.text(.1,1.4,'Residual block',fontsize=7.4,fontweight='semibold',color=INK)
node(net,3.3,.45,2.75,.75,'GN → SiLU\n3 × 3 conv',BLUE,size=7)
node(net,6.9,.45,2.55,.75,'GN → FiLM\nSiLU',TEAL,size=7)
node(net,10.3,.45,2.5,.75,'3 × 3 conv\n+ residual skip',BLUE,size=7)
link(net,(2.45,.825),(3.3,.825))
link(net,(6.05,.825),(6.9,.825))
link(net,(9.45,.825),(10.3,.825))
link(net,(12.8,.825),(13.65,.825))
net.text(2.15,.825,'$h$',ha='center',va='center',fontsize=8,color=INK)
net.text(14.6,.825,'To next block',ha='center',va='center',fontsize=7,color=INK)
# Routed paths around the architecture, to the inset's modulation inputs.
net.plot([2.9,3.15,3.15,8.65],[4.85,4.85,1.6,1.6],color='#009E73',lw=.8,zorder=1)
link(net,(8.65,1.6),(8.65,1.2),color='#009E73')
net.plot([2.65,3.05,3.05,7.65],[2.9,2.65,1.4,1.4],
         color='#8856A7',lw=.8,zorder=1)
link(net,(7.65,1.4),(7.65,1.2),color='#8856A7')
net.text(8.1,.05,r'FiLM: $(1+\gamma_t+\gamma_c)\,\mathrm{GN}(h)+\beta_t+\beta_c$',
         ha='center',fontsize=6.8,color=INK)

# (c) The same learned U-Net is called at every sampling transition.
sample.text(0,2.13,'(c)  Generation',fontsize=9,fontweight='semibold',color=INK)
node(sample,.1,.45,2.5,.85,'Gaussian noise\n205 × 205',GRAY)
node(sample,3.6,.45,3.2,.85,'Conditional U-Net (b)\nVelocity prediction',BLUE)
node(sample,7.8,.45,3.15,.85,'DDIM update\n50 steps; $\eta=0$',PURPLE)
node(sample,12.1,.45,3.8,.85,'Inverse asinh transform\nF115W estimate · MJy/sr',TEAL,size=7.1)
for start,end in [((2.6,.875),(3.6,.875)),((6.8,.875),(7.8,.875)),
                  ((10.95,.875),(12.1,.875))]:link(sample,start,end)
sample.text(5.2,1.98,'NISP + validity, $c$',ha='center',fontsize=7.2,color='#007C59')
link(sample,(5.2,1.84),(5.2,1.3),color='#007C59')
sample.plot([9.4,9.4,5.2],[.45,.12,.12],color=INK,lw=.8,zorder=1)
link(sample,(5.2,.12),(5.2,.45))
sample.text(7.3,.15,'next timestep',ha='center',va='bottom',fontsize=6.4,
            color=INK,bbox=dict(facecolor='white',edgecolor='none',pad=.5))
sample.text(11.5,1.07,'final',ha='center',fontsize=6.3,color=INK)
plt.show()
```

## Working caption

Conditional diffusion architecture. (a) Noisy JWST targets train the network
with a masked velocity-prediction loss and an image-reconstruction term.
(b) The U-Net uses three encoder–decoder levels with concatenated skip features
and two conditioned residual blocks per stage. NISP intensity, its validity
mask, and timestep embeddings modulate every residual block through FiLM
(feature-wise linear modulation). Labels give spatial dimensions and channels;
reflection padding and cropping preserve the 205 × 205 output grid.
(c) Generation starts from Gaussian noise and uses 50 DDIM steps, conditioned
on NISP and its validity mask. The output is transformed back to linear surface
brightness. GN denotes group normalization.

## Implementation notes

- NISP intensity is bilinearly upsampled; the input validity mask uses nearest
  neighbors. A learned conditioning pyramid follows the encoder and decoder.
- The 256-dimensional sinusoidal/MLP timestep embedding feeds all 14 residual blocks.
- Both input and target use saved asinh model coordinates. Target masks apply
  to both losses; the image L1 term is in model coordinates, not MJy/sr.
- Training samples 1,000 cosine-schedule timesteps with exactly zero terminal
  signal. Sampling uses 50 deterministic DDIM transitions, with no output clipping.
- Stage dimensions and residual-block operations were read from the sources
  whose hashes match the recorded full run. No model was constructed for this figure.
